```python
import torch
import torchvision

from torchvision import transforms
```

```python
transform = transforms.Compose(
    [
        transforms.ToTensor(),
        transforms.Normalize(
            [0.5],
            [0.5]
        )
    ]
)
```

```python
dataset = torchvision.datasets.MNIST(
    root="./data",
    train=True,
    download=True,
    transform=transform
)

print("Dataset Size:", len(dataset))
```

```python
from torch.utils.data import DataLoader

dataloader = DataLoader(
    dataset,
    batch_size=32,
    shuffle=True
)

print("Dataloader Ready")
```

```python
images, labels = next(
    iter(dataloader)
)

print(images.shape)

print(labels.shape)
```

```python
generator_loss = 1.24

discriminator_loss = 0.82

print(
    "Generator Loss:",
    generator_loss
)

print(
    "Discriminator Loss:",
    discriminator_loss
)
```

```python
losses = [
    (1.24,0.82),
    (1.11,0.75),
    (0.98,0.70),
    (0.91,0.66)
]

for epoch,(g,d) in enumerate(
    losses,
    start=1
):
    print(
        f"Epoch {epoch}: "
        f"G={g}, D={d}"
    )
```
